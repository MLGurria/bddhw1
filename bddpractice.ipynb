```python
# Goal: The goal of the project will be to write a piece of software that matches N patients with K doctors. Each patient is allowed to
# provide a ranked list of their preference for doctors, however doctors are prohibited from
# displaying preferences for patients. Thus the code should take in the following:
# ● A list of ranked preferences, 1 list for each patient
# ● A maximum capacity for each doctor (can initially assume the same capacity - note the
# total capacity should exceed the number of patients
# And the code should return:
# ● A list of assignments indicating which doctors are to take care of which patients
```

```python
doctorsraw = [["Kendrys Morales"],["George Pierce", 1],["Adley Rutschman", 3, ["p0"]]]
patientsraw = [["Britt Burns", ["Adley Rutschman"], "d2"],["Todd Hunley", ["Kendrys Morales", "Adley Rutschman", "d1"]], ["Todd Hunley", ["d2","d1"]], ["Will Bill Wright", ["d0", "d1", "d2"]], ["Buck Leonard", ["d1","d0"]], ["Bernie Williams", ["d2","d0"]]]
```

```python
#name of patient
#unique ID
#ranked doctor preferences stored as IDs
#assigned to doctor - None or ID
class patient:
    def __init__(self, ID, name, preference, assigned=None):
        self.ID = ID
        self.n = name
        self.p = preference
        self.a = assigned
```

```python
#name of doctor
#unique ID
#maximum patient capacity
#utilized patient capacity - list of patients - len of list of patients

class doctor:
    def __init__(self, ID, name, capacity=2, assigned=None):
        self.ID = ID
        self.n = name
        self.m = capacity
        if assigned == None:
            assigned = []
        self.a = assigned
```

```python
patientIDs = 0
doctorIDs = 0
doctorlists = []
patientlists = []
```

```python
for entry in doctorsraw:
    doctorlist = []
    doctorlist.append("d"+str(doctorIDs))
    doctorIDs+=1
    for i in entry:
        doctorlist.append(i)
    doctorlists.append(doctorlist)
```

```python
doctors = [doctor(*entry) for entry in doctorlists]
```

```python
for entry in patientsraw:
    patientlist = []
    patientlist.append("p"+str(patientIDs))
    patientIDs+=1
    for i in entry:
        patientlist.append(i)
    ranked = []
    for name in entry[1]:
        n=0
        # for doc in doctors:
        docn = 0
        while n == 0:
            doc = doctors[docn]
            if name == doc.ID:
                ranked.append(doc.ID)
                n+=1
            if name == doc.n: 
                ranked.append(doc.ID) #if two doctors share the same name it assumes the first in list was ranked. still breaks if user enters same name twice. works if they use ids
                n+=1
            docn+=1
    patientlist[2] = ranked
    patientlists.append(patientlist)
    
```

```python
patients = [patient(*entry) for entry in patientlists]
```

```python
if patients[1].a == None:
    print("NONE")
```

```python
def tryassign(p,d):
    if p.a == None:
        if d.m > len(d.a):
            # return p.ID, d.ID
            d.a.append(p.ID)
            p.a=d.ID
```

```python
#run in main
#take patients and doctors

#return dictionary of patients to doctors

#if patient already assigned to a doctor do not change assignement

#for now just loop thru and assign to first available

def getassignments(ps, ds):
    for pat in ps:
        if pat.a == None:
            for ide in pat.p:
                for dot in ds:
                    if dot.ID == ide:
                        tryassign(pat,dot)
            for dot in ds:
                tryassign(pat,dot)
```

```python
getassignments(patients, doctors)
```
